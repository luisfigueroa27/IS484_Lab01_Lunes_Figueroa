<b>Parte A:</b> Completar la ficha del problema para el siguiente caso: un repartidor debe ir desde el
Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

Estado | Descripcion
---|---
Estado inicial | El repartidor se encuentra en el almacen
Estado objetivo | el repartidor ha llegado a ZONA N0RTE
Acciones posibles | El repartidor se puede desplazar en cualquier punto cercano a el
Costo de cada accion | Consideramos el valor de 1 para cualquier tramo

<b>Parte B:</b> Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de
listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

<h4>Implementación del codigo de BFS</h4>

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]] # cola de caminos, empieza con el estado inicial
    visitados = set()
    while cola:
        camino = cola.pop(0) # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]
        
        if nodo_actual == objetivo:
            return camino
            
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
    
    return None # no se encontro un camino
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

<h3>2.1 Planteamiento del problema</h3>

Se quiere encontrar, dentro de un conjunto de observaciones ya conocidas, cuál es la más parecida a
una observación nueva

<b>Parte A:</b> Implementación de DFS <br>
Completar la búsqueda en profundidad sobre el mismo mapa de la Sección 4.2, usando una pila
(lista de Python, con .pop() para sacar el último elemento):

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]]
    visitados = set()
    while pila:
        camino = pila.pop() # sacar el ULTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]
        if nodo_actual == objetivo:
            return camino
            
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
        
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: misma logica que en BFS, pero agregando a la pila
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)
    return None
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

<b>Parte B:</b> Encontrar la observación más cercana <br>
Dado el dataset de la Sección 4.4 (cuatro observaciones de temperatura/humedad) y una observación
nueva, calcular la distancia entre la nueva observación y cada una de las existentes, e identicar cuál
es la más cercana:

```python
nueva_observacion = np.array([20.5, 48])

distancias = []
for i, obs in enumerate(dataset):
    # COMPLETAR: calcular la distancia euclidiana entre 'nueva_observacion' y 'obs'
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observacion {i}: {d}")

indice_mas_cercano = np.argmin(distancias)
print("La observacion mas parecida es la numero", indice_mas_cercano)
```

<B>Parte C:</B> Reflexión

<B>¿Por qué la distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones?</B>

La distancia euclidiana es una forma razonable de medir la similitud porque permite calcular qué tan cerca o lejos están dos observaciones considerando sus características. Si la distancia es pequeña, significa que los valores de las características son similares; si la distancia es grande, las observaciones son más diferentes. En este laboratorio se utiliza para comparar observaciones de temperatura y humedad, siguiendo el principio de que cuanto menor sea la distancia, más parecidas son las observaciones. Este concepto es utilizado posteriormente en algoritmos como KNN y K-Means.
